## 1. Load Weather Dataset - CSV(in).csv file


```python
import sqlite3
import pandas as pd
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
```

```python
spark = SparkSession.builder.appName("Daily_Weather_Observations").getOrCreate()
weather_dataset_df = spark.read.csv("Weather Dataset - CSV(in).csv", header=True, inferSchema=True)
weather_dataset_df.show(5)
weather_dataset_df.printSchema()
```

## 2. Task 1

Write a function that returns the number of days when it rained the next day. Hint: Count the number of rows where RainTomorrow is ‘Yes’. 


```python
def count_rainy_tomorrows(df):
    return df.filter(F.col("RainTomorrow") == "Yes").count()

print(count_rainy_tomorrows(weather_dataset_df))
```

## 3. Task 2

Write a function that returns the average sunshine duration on days with no rainfall. Hint: Filter rows where Rainfall is 0 and then calculate the average of Sunshine. 


```python
def average_sunshine_duration_with_no_rainfall(df):
    no_rain_df = df.filter(F.col("Rainfall") == 0)
    return no_rain_df.agg(F.avg("Sunshine")).collect()[0][0]

print(average_sunshine_duration_with_no_rainfall(weather_dataset_df))
```

## 4. Task 3

Write a function that returns the maximum temperature recorded at 3 PM. Hint: Use the Temp3pm column and return the max value. 


```python
def max_temperature_at_3pm(df):
    return df.agg(F.max("Temp3pm")).collect()[0][0]

print(max_temperature_at_3pm(weather_dataset_df))
```

### 5. Task 4

Write a function that returns the average humidity at 3 PM on days it rained the next day. Hint: Filter rows where RainTomorrow is ‘Yes’ and then calculate the mean of Humidity3pm.


```python
def avg_humidity_at_3pm_rained_next_day(df):
    rained_next_day_df = df.filter(F.col("RainTomorrow") == "Yes")
    return rained_next_day_df.agg(F.avg("Humidity3pm")).collect()[0][0]

print(avg_humidity_at_3pm_rained_next_day(weather_dataset_df))
```

### 6. Task 5

Write a function that returns the most common wind direction at 9 AM on cloudy days. Hint: Filter rows where Cloud9am is above a threshold (e.g. >5) and then find the mode of WindDir9am.


```python
def most_common_wind_direction_at_9am_on_cloudy_days(df):
    cloudy_days_df = df.filter(F.col("Cloud9am") > 5)
    return cloudy_days_df.agg(F.mode("WindDir9am")).collect()[0][0]

print(most_common_wind_direction_at_9am_on_cloudy_days(weather_dataset_df))
```
